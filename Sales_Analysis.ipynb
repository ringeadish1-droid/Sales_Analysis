```python
x1 =  ["#E65100", "#EF6C00",]

plt.figure(figsize=(15,6))


A1 = sns.countplot(x="Zone",data=df,hue="Marital_Status",palette= x1,edgecolor ="black")
for i in A1.containers:
    A1.bar_label(i)


plt.title("Marital Status Wise Age Group Count",color = "red",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
x3 = ["#264653", "#2F5D62", "#3D7A80", "#5B9EA6", "#83BBC0", "#AED5D8"]

plt.figure(figsize=(15,6))

A2 = sns.countplot(x='Age Group',data=df,hue='Marital_Status',palette=x3)


for i in A2.containers:
    A2.bar_label(i)

plt.title("Marital Status Wise Age Group Count",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python


Linechart = df.groupby("Sector",as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)

Linechart
```

```python
plt.figure(figsize=(15,6))

sns.lineplot(x="Sector",y="Amount",data=Linechart,marker="o")

plt.xticks(rotation=90)

plt.grid()

plt.title("Sector Wise Total Amount in Sorted Format")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()
```

```python
chart_2 = df.groupby("State",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
```

```python
x2 = ["#37474F", "#455A64", "#546E7A", "#78909C", "#90A4AE", "#B0BEC5","#F57C00", "#FF9800", "#FFB74D", "#FFD180"]
plt.figure(figsize=(15,6))

A2 = sns.barplot(x="State",y="Orders",data=chart_2,edgecolor='black',palette=x2)
for i in A2.containers:   # Count Bar Range 
    A2.bar_label(i)

plt.xticks(rotation=90)

plt.title("Top 10 State Order Wise",color='red',fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

Pie Chart

```python
piechart = df.groupby("Zone",as_index=False)["User_ID"].count()

plt.pie(piechart["User_ID"],labels=piechart["Zone"],autopct="%.f%%")

plt.title("Zone Wise Total Users",fontweight="bold", fontsize=16, family="sans-serif")


plt.show()
```
